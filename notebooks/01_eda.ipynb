```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


ROOT = Path.cwd().parent
movies_clean_file = ROOT / "data" / "clean" / "movies_clean.csv"

df = pd.read_csv(movies_clean_file)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
plt.Figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="vote_average",
    bins=20,
    kde=True,
)
plt.title("Distribution of movie ratings")
plt.xlabel("Vote average")
plt.ylabel("Number of movies")
plt.show()
```

Most movies have an average rating between 6 and 8, with the highest concentration around 6.5–7.5. Very low ratings are rare, and there are only a few movies with ratings close to 10.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="popularity",
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Movies")

plt.show()
```

Movie popularity is highly right-skewed. Most movies have relatively low popularity values, while a small number of movies have very high popularity creating the long tail on the right
This means popularity is not evenly distributed across movies.a

```python
import ast

import ast

df["genres"] = df["genres"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)
genre_counts = df["genres"].explode().apply(lambda x: x["name"] if isinstance(x, dict) else None).value_counts()


plt.figure(figsize=(10, 6))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title("Movies Genres")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")

plt.show()
```

Drama, Action, and Comedy are the most represented genres in the dataset. Thriller and Adventure are also very common, while genres such as History and War appear much less frequently.

```python
df["release_date"].dtype
```

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
movies_per_year = df["release_date"].dt.year.value_counts().sort_index()

plt.figure(figsize=(12, 5))

sns.lineplot(
    x=movies_per_year.index,
    y=movies_per_year.values
)

plt.title("Movie Releases by Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")

plt.show()
```

Movie releases increase strongly over time, especially from the 1980s onward. The dataset contains many more recent movies than older ones, with a very large peak in the most recent years. This suggests the TMDB sample is heavily concentrated on newer releases.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="runtime",
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Runtime")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

Most movies have a runtime between about 90 and 130 minutes, with the highest concentration around 90–120 minutes. Very short and very long movies are much less common.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="budget",
    y="revenue"
)

plt.title("Budget vs Revenue")
plt.xlabel("Budget")
plt.ylabel("Revenue")

plt.show()
```

There is a general positive relationship between budget and revenue: movies with larger budgets often generate higher revenue. However, the relationship is not perfect, since movies with similar budgets can have very different revenues and there are several outliers.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="vote_count",
    y="popularity"
)

plt.title("Vote Count vs Popularity")
plt.xlabel("Vote Count")
plt.ylabel("Popularity")

plt.show()
```

There is no strong linear relationship between vote count and popularity. Most movies have relatively low popularity even when they have many votes, while a few movies with few votes have very high popularity. The data also contains several strong outliers.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="vote_average"
)

plt.title("Boxplot of Movie Ratings")
plt.ylabel("Vote Average")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="runtime"
)

plt.title("Boxplot of Movie Runtime")
plt.ylabel("Runtime (minutes)")

plt.show()
```

```python
corr = df[
    [
        "runtime",
        "budget",
        "revenue",
        "popularity",
        "vote_average",
        "vote_count",
    ]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

The strongest correlations are between budget and revenue (0.66) and between revenue and vote_count (0.60). Most other relationships are weak to moderate. Popularity has relatively low correlation with the other numerical variables.
